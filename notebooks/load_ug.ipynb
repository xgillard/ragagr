```python
from langchain_pymupdf4llm import PyMuPDF4LLMLoader
from langchain_text_splitters import MarkdownHeaderTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
loader = PyMuPDF4LLMLoader(
    file_path="https://agatha.arch.be/help/States-Archives_user-guide-genealogical-sources.pdf",
    mode="single",
    #extract_images=False, # on ne veut pas etre pollué par des données d'exemple
    #images_parser= TesseractBlobParser(langs=("eng", "fra", "nld")),
)
splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on= [
        ("##",  "h2"),
        ("###", "h3"),
    ],
    strip_headers=False,
)
docs   = loader.load()
chunks = splitter.split_text(docs[0].page_content)
chunks = chunks[5:] # osef de la table des matieres

chroma = Chroma(
    "genealogy", 
    HuggingFaceEmbeddings(
        model_name   = "intfloat/multilingual-e5-large-instruct",
        model_kwargs = {"device": "cuda", "model_kwargs": {"torch_dtype": "float16"}, },
    ),
    persist_directory="./digita",
)

chroma.add_documents(chunks)
```
